9. 第 217 题：从空白定义 contains_duplicate(nums)，使用集合解决。
[2,5,2] → True；[2,5,7] → False；[7] → False；[1,1] → True。
不看参考实现，解释为什么要先检查再 add，return False 为什么在循环外。


```python
# 用 set() 创建空集合，然后写循环及四组测试
seen = set() # set()：创建空集合
seen.add(2)   # add：将数字加入集合

print(2 in seen)   # True，2 已在集合中
print(5 in seen)   # False，5 还没有出现

def y(nums):   # 判断列表是否包含重复值
    seen = set()   # 保存此前见过的数字
    for x in nums :    # 依次读取数字
        if x in seen :      # 先检查这个数字是否已经出现过
            return True      # 出现过，说明存在重复
        seen.add(x)        # 第一次见到，加入集合

        
    return False        # 全部检查完没有重复；这句在循环外


print(y([2,5,2]))
print(y([2, 5, 7]))  # False
print(y([7]))        # False
print(y([1, 1])) 
```
